```python
# Step 1: inspect columns and dtypes
import pandas as pd

print("Columns:\n", qm9.columns.tolist())
print("\nDtypes:\n", qm9.dtypes)
```

```python
# Step 2: robust value_counts for a charge-like column (or fallback to best categorical)
import numpy as np

# find charge-like column
charge_col = next((c for c in qm9.columns if 'charge' in c.lower() or 'chg' in c.lower() or 'formal' in c.lower()), None)

if charge_col is None:
    # fallback candidates: object dtype or low-cardinality numeric
    candidates = [c for c in qm9.columns if (qm9[c].dtype == 'object') or (qm9[c].nunique() <= 50)]
    print("No explicit 'charge' column found. Candidates:", candidates[:10])
    charge_col = candidates[0] if candidates else None

if charge_col is None:
    print("No suitable categorical column found to run value_counts(). Inspect qm9.columns manually.")
else:
    print("Using column for counts:", charge_col)
    # handle unhashable entries (lists/arrays) safely
    try:
        vc = qm9[charge_col].value_counts(dropna=False)
    except Exception:
        safe_series = qm9[charge_col].apply(lambda x: tuple(x) if isinstance(x, (list, tuple, np.ndarray)) else x)
        vc = safe_series.value_counts(dropna=False)

    print("\nTop value counts:\n", vc.head(20))
    print("\nPercentages:\n", (vc / len(qm9) * 100).round(2).head(20))
```

```python
# Create a new categorical feature: SMILES length
qm9['smiles_length'] = qm9['smiles'].str.len()

# Value counts (categorical-like)
print(qm9['smiles_length'].value_counts())
print(qm9['smiles_length'].value_counts(normalize=True)*100)
```

```python
# Create derived categorical column: HOMO sign
qm9['HOMO_sign'] = qm9['homo'].apply(lambda x: 'Positive' if x > 0 else 'Negative')

# Crosstab between SMILES length and HOMO sign
pd.crosstab(qm9['smiles_length'], qm9['HOMO_sign'])
```

```python
# Step 4: find homo/lumo columns (common names 'homo' and 'lumo') and preview
homo_col = next((c for c in qm9.columns if c.lower() == 'homo' or 'homo' in c.lower()), None)
lumo_col = next((c for c in qm9.columns if c.lower() == 'lumo' or 'lumo' in c.lower()), None)

print("homo_col =", homo_col, "  lumo_col =", lumo_col)

if homo_col and lumo_col:
    print(qm9[[homo_col, lumo_col]].head(8))
else:
    print("HOMO/LUMO columns not found exactly; run print(qm9.columns.tolist()) to inspect names.")
```

```python
# Step 5: sort by homo (ascending and descending) — safe only if homo_col exists
if homo_col:
    print("Lowest 5 by HOMO:")
    print(qm9[[homo_col, lumo_col]].sort_values(homo_col).head(5))
    print("\nHighest 5 by HOMO:")
    print(qm9[[homo_col, lumo_col]].sort_values(homo_col, ascending=False).head(5))
else:
    print("Skipping sort: homo_col not found.")
```

```python
# Step 6: create energy gap column (LUMO - HOMO), convert to numeric safely
if homo_col and lumo_col:
    qm9['gap'] = pd.to_numeric(qm9[lumo_col], errors='coerce') - pd.to_numeric(qm9[homo_col], errors='coerce')
    print("Created 'gap' column. Top 5 by gap:")
    print(qm9[[homo_col, lumo_col, 'gap']].sort_values('gap', ascending=False).head(5))
else:
    print("Cannot compute gap: missing homo or lumo column.")
```

```python
# Define homo_col
homo_col = 'homo'

# Groupby smiles_length (categorical) and calculate mean HOMO
if homo_col in qm9.columns and 'smiles_length' in qm9.columns:
    homo_by_smiles_len = qm9.groupby('smiles_length')[homo_col].mean().reset_index()
    print("Average HOMO by SMILES length:")
    print(homo_by_smiles_len)
else:
    print("Required columns not found")
```

```python
#Step 8 (Alternative): Groupby by number of atoms (A column) and average HOMO

if homo_col in qm9.columns and 'A' in qm9.columns:
    homo_by_atoms = qm9.groupby('A')[homo_col].mean().reset_index()
    print("Average HOMO by number of atoms (A):")
    print(homo_by_atoms)
else:
    print("Required columns not found")

```

```python
# Step 9: Merge-like comparison (simplified for QM9)
# Example: compare average HOMO and LUMO energies side by side
homo_avg = qm9['homo'].mean()
lumo_avg = qm9['lumo'].mean()

comparison = pd.DataFrame({
    'Property': ['HOMO Energy', 'LUMO Energy'],
    'Average Value': [homo_avg, lumo_avg]
})
print(comparison)
```
