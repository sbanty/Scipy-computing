```python
# Sorting
economy[['Year','GDP (current US$) ']].sort_values('GDP (current US$) ')[0:5]
economy[['Year','GDP (current US$) ']].sort_values('GDP (current US$) ', ascending=False)[0:5]
```

```python
# New column: GDP per capita difference (example)
economy['eco_gap'] = economy['GDP (current US$) '] - economy['Inflation, consumer prices (annual %)']
economy[['Year','GDP (current US$) ','Inflation, consumer prices (annual %)','eco_gap']].sort_values('eco_gap', ascending=False)[0:5]
```

```python
# Groupby mean GDP per decade
economy['Decade'] = (economy['Year'] // 10) * 10
economy.groupby('Decade')['GDP (current US$) '].mean()
```

```python
# Reset index
gdp_by_year = economy.groupby('Year')['GDP (current US$) '].mean().reset_index()
print(gdp_by_year)
```

```python
# Groupby with multiple keys
gdp_by_sector = economy.groupby(['Year','GDP growth (annual %)'])['GDP (current US$) '].mean().reset_index()
print(gdp_by_sector)
```

```python
# Merge
gdp_by_sector = economy.groupby(['Year','GDP growth (annual %)'])['GDP (current US$) '].mean().reset_index()
comparison = gdp_by_sector.merge(gdp_by_year, on='Year', how='outer')
comparison.rename(columns={'GDP (current US$) _x':'GDP_BY_YEAR_SECTOR','GDP (current US$) _y':'GDP_BY_YEAR'}, inplace=True)
comparison['change'] = comparison.apply(lambda rec:(rec.GDP_BY_YEAR_SECTOR - rec.GDP_BY_YEAR)/rec.GDP_BY_YEAR, axis=1)
comparison
```

```python
# Filter: high inflation years
economy[economy['Inflation, consumer prices (annual %)'] > 10][['Year','Inflation, consumer prices (annual %)']]
```

```python
# Drop unnecessary column if exists
economy.drop('Sl.NO.', inplace=True, axis=1, errors='ignore')
economy.columns
```

```python
# First, check actual column names in your dataset
print(economy.columns.tolist())

# Clean column names by stripping spaces
economy.columns = economy.columns.str.strip()

# Now rename cleanly
economy_renamed = economy.rename(columns={
    "GDP (current US$)": "GDP_USD",
    "Inflation, consumer prices (annual %)": "Inflation"
})

# Select features
influential_features = ["GDP_USD", "Inflation"]

# Pairplot
import seaborn as sn
sn.pairplot(economy_renamed[influential_features], height=2)

# Heatmap
import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
sn.heatmap(economy_renamed[influential_features].corr(), annot=True, cmap="coolwarm")
plt.show()
```

```python
# Matplotlib boxplot + whisker stats
box = plt.boxplot(economy['GDP (current US$) '])
[item.get_ydata()[0] for item in box['whiskers']]
[item.get_ydata()[0] for item in box['medians']]
plt.title('Distribution of GDP (current US$)')
plt.ylabel('GDP (current US$)')
```
