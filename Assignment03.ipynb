```python
# Robust filter: find real column names and then filter
# (works even if your column names contain special chars like "Avg_Temp (°C)")

# 1) auto-detect columns by substring
temp_cols = [c for c in climate.columns if 'temp' in c.lower()]
co2_cols  = [c for c in climate.columns if 'co2'  in c.lower()]
sea_cols  = [c for c in climate.columns if 'sea'  in c.lower() or 'sea_level' in c.lower()]

print("Temp-like columns found:", temp_cols)
print("CO2-like columns found: ", co2_cols)
print("Sea-like columns found: ", sea_cols)

# 2) choose the first matching column of each type (if any)
chosen_temp = temp_cols[0] if temp_cols else None
chosen_co2  = co2_cols[0]  if co2_cols  else None
chosen_sea  = sea_cols[0]  if sea_cols  else None

# 3) make sure Year/Month exist
if 'Year' not in climate.columns or 'Month' not in climate.columns:
    print("Warning: 'Year' or 'Month' column not found. Available columns:\n", list(climate.columns))
else:
    # 4) build selected columns list (only include columns that actually exist)
    sel = ['Year','Month']
    if chosen_temp: sel.append(chosen_temp)
    if chosen_co2:  sel.append(chosen_co2)
    if chosen_sea:  sel.append(chosen_sea)

    # 5) final filter: only run if we found a temperature-like column
    if not chosen_temp:
        print("No temperature-like column found. Run print(list(climate.columns)) to inspect names.")
    else:
        # choose a threshold (25 deg) — you can change this if needed
        thresh = 25
        # ensure numeric (coerce errors already done earlier; keep .dropna() safe)
        filtered = climate[climate[chosen_temp] > thresh][sel]
        # show results (or a message if empty)
        if filtered.empty:
            print(f"No rows where {chosen_temp} > {thresh}. Try lowering threshold or inspect data.")
            display(climate[sel].head(10))
        else:
            display(filtered.reset_index(drop=True))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Make a safe copy of needed columns
plot_df = climate_df[['Avg_Temp (°C)', 'Sea_Surface_Temp (°C)']].copy()

# Force numeric conversion (turn invalid text into NaN)
plot_df['Avg_Temp (°C)'] = pd.to_numeric(plot_df['Avg_Temp (°C)'], errors='coerce')
plot_df['Sea_Surface_Temp (°C)'] = pd.to_numeric(plot_df['Sea_Surface_Temp (°C)'], errors='coerce')

# Drop rows with missing values
plot_df = plot_df.dropna()

# Scatter with regression line
plt.figure(figsize=(8,6))
sns.regplot(x='Avg_Temp (°C)', y='Sea_Surface_Temp (°C)', data=plot_df)
plt.title("Avg Temp vs Sea Surface Temp (with Trend Line)")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Copy relevant columns
plot_df = climate_df[['CO2_Concentration (ppm)', 'Avg_Temp (°C)']].copy()

# Convert to numeric (handles strings/NaN safely)
plot_df['CO2_Concentration (ppm)'] = pd.to_numeric(plot_df['CO2_Concentration (ppm)'], errors='coerce')
plot_df['Avg_Temp (°C)'] = pd.to_numeric(plot_df['Avg_Temp (°C)'], errors='coerce')

# Drop missing
plot_df = plot_df.dropna()

# Plot regression
plt.figure(figsize=(8,6))
sns.regplot(x='CO2_Concentration (ppm)', y='Avg_Temp (°C)', data=plot_df)
plt.title("CO2 Concentration vs Avg Temperature")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Select important indicators
influential_features = ['Avg_Temp (°C)', 'CO2_Concentration (ppm)', 'Sea_Surface_Temp (°C)']

# Copy and clean data
plot_df = climate_df[influential_features].copy()
for col in influential_features:
    plot_df[col] = pd.to_numeric(plot_df[col], errors='coerce')
plot_df = plot_df.dropna()

# Pairplot — relationships
sns.pairplot(plot_df, height=2)
plt.show()

# Heatmap — correlations
sns.heatmap(plot_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation between Climate Indicators")
plt.show()
```
