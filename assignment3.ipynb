```python
# Crosstab — between Type and Al content bucketed
pd.crosstab(pd.cut(glass['Al'], bins=5), glass['Type'])
```

```python
# Sorting
glass[['Id','RI']].sort_values('RI')[0:5]
glass[['Id','RI']].sort_values('RI', ascending=False)[0:5]
```

```python
# New column: Difference between Na and Mg
glass['Na_Mg_diff'] = glass['Na'] - glass['Mg']
glass[['Id','Na','Mg','Na_Mg_diff']].sort_values('Na_Mg_diff', ascending=False)[0:5]
```

```python
# Groupby mean RI by Type
glass.groupby('Type')['RI'].mean()
```

```python
# Groupby with reset
ri_by_type = glass.groupby('Type')['RI'].mean().reset_index()
print(ri_by_type)
```

```python
# Groupby with multiple keys (Type + Mg bucket)
ri_by_type_mg = glass.groupby(['Type', pd.cut(glass['Mg'], bins=4)])['RI'].mean().reset_index()
print(ri_by_type_mg)
```

```python
# Merge comparison
comparison = ri_by_type_mg.merge(ri_by_type, on='Type', how='outer')
comparison.rename(columns={'RI_x':'RI_TYPE_MG','RI_y':'RI_TYPE'}, inplace=True)
comparison['change'] = comparison.apply(lambda rec:(rec.RI_TYPE_MG - rec.RI_TYPE)/rec.RI_TYPE, axis=1)
comparison
```

```python
# Filter example — high Si content
glass[glass['Si'] > 73][['Id','Si','Type']]
```

```python
# Drop column example
glass.drop('Id', inplace=True, axis=1, errors='ignore')
glass.columns
```

```python
# Filter with isin: show only types 1 and 2
glass[glass['Type'].isin([1, 2])]

# Filter with between: RI between 1.515 and 1.520
glass[glass['RI'].between(1.515, 1.520)]
```

```python
# Example: merge average Na and average Mg by Type
avg_na = glass.groupby('Type')['Na'].mean().reset_index(name='Avg_Na')
avg_mg = glass.groupby('Type')['Mg'].mean().reset_index(name='Avg_Mg')

merged = pd.merge(avg_na, avg_mg, on='Type', how='inner')
merged
```

```python
# Filter with isin: show only glass types 1 and 2
glass[glass['Type'].isin([1, 2])][['RI','Na','Mg','Type']]

# Filter with between: RI between 1.515 and 1.520
glass[glass['RI'].between(1.515, 1.520)][['RI','Type']]
```

```python
# Merge average Na and Mg by glass type
avg_na = glass.groupby('Type')['Na'].mean().reset_index(name='Avg_Na')
avg_mg = glass.groupby('Type')['Mg'].mean().reset_index(name='Avg_Mg')

merged = pd.merge(avg_na, avg_mg, on='Type', how='inner')
print("Merged averages of Na and Mg by glass type:")
print(merged)
```
