```python
import pandas as pd
from google.colab import files

# Load directly (since you already have winequality-red.csv in your environment)
# Upload the file
uploaded = files.upload()

# Read the uploaded file using the correct file name
file_name = list(uploaded.keys())[0]
wine = pd.read_csv(file_name, sep=";")

# Check top rows
print(wine.head())
```

```python
# Value counts
wine['quality'].value_counts()
wine['quality'].value_counts(normalize=True)*100
```

```python
# Crosstab between quality and alcohol bins
pd.crosstab(pd.cut(wine['alcohol'], bins=5), wine['quality'])
```

```python
# Sorting
wine[['alcohol','quality']].sort_values('alcohol')[0:5]
wine[['alcohol','quality']].sort_values('alcohol', ascending=False)[0:5]
```

```python
# New column: "acidity_ratio" = fixed acidity / volatile acidity
wine['acidity_ratio'] = wine['fixed acidity'] / wine['volatile acidity']
wine[['fixed acidity','volatile acidity','acidity_ratio']].head()
```

```python
# Groupby mean
wine.groupby('quality')['alcohol'].mean()
```

```python
# Groupby with reset
alcohol_by_quality = wine.groupby('quality')['alcohol'].mean().reset_index()
print(alcohol_by_quality)
```

```python
# Groupby with multiple keys
grouped = wine.groupby(['quality','pH'])['alcohol'].mean().reset_index()
print(grouped.head())
```

```python
# Merge example
comparison = grouped.merge(alcohol_by_quality, on='quality', how='outer')
comparison.rename(columns={'alcohol_x':'ALCOHOL_BY_QUALITY_PH','alcohol_y':'ALCOHOL_BY_QUALITY'}, inplace=True)
comparison['change'] = comparison.apply(lambda rec:(rec.ALCOHOL_BY_QUALITY_PH - rec.ALCOHOL_BY_QUALITY)/rec.ALCOHOL_BY_QUALITY, axis=1)
comparison.head()
```

```python
# Filter example: high alcohol wines
wine[wine['alcohol'] > 12][['alcohol','quality']]
```

```python
import matplotlib.pyplot as plt
import seaborn as sn
%matplotlib inline
```

```python
# Barplot
sn.barplot(x='quality', y='alcohol', data=wine)
```

```python
# Histogram
plt.hist(wine['alcohol'], bins=20)
plt.xlabel("Alcohol")
plt.ylabel("Count")
plt.title("Distribution of Alcohol Content")
plt.show()
```

```python
# Boxplot
sn.boxplot(x='quality', y='alcohol', data=wine)
```

```python
import seaborn as sn
import matplotlib.pyplot as plt

# Influential features
influential_features = ['fixed acidity','volatile acidity','citric acid','alcohol','quality']

# Pairplot
sn.pairplot(wine[influential_features], height=2)
plt.suptitle("Pairplot of Influential Features", y=1.02)  # Title above the pairplot
plt.show()  # <-- separates the plots

# Heatmap
plt.figure(figsize=(8,6))
sn.heatmap(wine[influential_features].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Influential Features")
plt.show()

```
